# Training-set model comparison for AChE

Compare a small fixed set of regressors using five-fold molecule-group cross-validation within the training portion of the same reproducible split. The held-out molecules are not used to rank models.

This replaces environment-dependent LazyPredict errors and plots of training-fit performance with an explicit, reproducible evaluation.

```python
from pathlib import Path
import sys

PROJECT = Path.cwd()
if not (PROJECT / "qsar_workflow.py").is_file():
    PROJECT = PROJECT / "computational_Drug_Discovery"
if not (PROJECT / "qsar_workflow.py").is_file():
    raise FileNotFoundError("Open this notebook from the project folder or repository root.")
if str(PROJECT) not in sys.path:
    sys.path.insert(0, str(PROJECT))

from qsar_workflow import acquire_data, prepare_descriptors, compare_regressors
records = acquire_data()
table, feature_columns = prepare_descriptors(records)
comparison = compare_regressors(table, feature_columns)
```

## Reading the table

MAE_mean and RMSE_mean are means of validation-fold errors; lower is better. R2_mean is the mean validation-fold R2; higher is better. The *_sd columns describe variation across five folds and are not confidence intervals. Preprocessing is fitted separately within each fold.

This ranking is exploratory. If it motivates choosing or tuning a different final model, retain a genuinely untouched evaluation set for that subsequent model. Do not reinterpret the already inspected baseline holdout as independent confirmation of every model choice.

The table is saved to outputs/training_group_cv_metrics.csv. Its feature representation is recorded in outputs/feature_columns.json.
