## Optional PaDEL fingerprints

The exported outputs/ache_molecules.smi contains SMILES followed by the unique record_id. Configure PaDEL so its Name column preserves that ID. After computing descriptors with a separately installed and documented PaDEL version, set ACHE_PADEL_CSV to the output CSV before running this notebook or pass its path as padel_csv to prepare_descriptors.

Missing, extra or duplicated descriptor IDs cause an error. This prevents a reordered descriptor file from assigning one compound's activity to another compound. Re-run both modelling notebooks after changing the feature representation.

## Outputs and predictor scope

outputs/ache_model_dataset.csv contains the aligned model table. outputs/feature_columns.json records the actual feature representation. Identifiers, source indices, activity labels and pIC50 are never predictor columns. The source and predictor metadata must accompany any reported model results.
