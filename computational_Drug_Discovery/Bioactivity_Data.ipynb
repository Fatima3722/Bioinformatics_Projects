# Acetylcholinesterase dataset and provenance

This educational workflow uses one immutable DataProfessor AChE archive in every notebook. It does not query a different GPCR target or select a target by its position in search results.

The archive already supplies pIC50 and four molecular descriptors. It does not retain the original target/species, assay, measurement-relation or ChEMBL-release metadata, so it cannot establish per-record assay comparability or a freshly curated human-only dataset.

Data and tutorial credit: [DataProfessor](https://github.com/dataprofessor/data), commit 421ee9d4e2f9c4d7ec4bccbf73c19f773674430a. See README.md for scope and provenance.

## Classification and record identity

pIC50 is supplied by the archived dataset. Labels are recomputed consistently: pIC50 >= 6 is active, pIC50 <= 5 is inactive, and the interval between them is intermediate. These are fixed educational thresholds, not clinical classifications. Original labels are retained for audit.

A unique workflow record_id is retained throughout the analysis. It identifies a workflow record, not an original ChEMBL activity measurement.

```python
columns = ['record_id', 'molecule_chembl_id', 'bioactivity_class', 'pIC50']
print(records[columns].head().to_string(index=False))
print('Unique molecules:', records.molecule_chembl_id.nunique())
print('Unique records:', records.record_id.nunique())
assert records.record_id.is_unique
```
